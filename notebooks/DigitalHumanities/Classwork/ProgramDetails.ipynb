Create a chart showing who puts in the most effort. number of hours (avg)

```python
import pandas as pd 
df = pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="pie",  x="Student", y="Hours_Studied", labels=df["Student"], legend=False, autopct="%1.1f%%") #, y="Hours_Studied", labels=df["Student"], legend=False, autopct="%1.1f%%"
plt.title("Average hours studied")
plt.show
```

```python
import matplotlib.pyplot as plt
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", color="#D9A0F2")
plt.show
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Grade"], alpha=0.6, color="#D9A0F2") #alpha for trasnparecy, if there was more data and overlap
#s=df["Years_of_practice"]*60 means if a studen haa more years of practice their bubble is bigger
plt.title("Hours studied vs grade")
plt.xlabel("Hours")
plt.ylabel("Grade")
plt.show()

```
