```python
# can use to work with tabular data
import pandas as pd 
df = pd.read_csv("Competition.csv")
df.head()# show the first few rows
df.describe() #show some basic statistics
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visuals
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of practice per person")
plt.ylabel("Years of prctice")
plt.show()
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visuals
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age of each person")
plt.ylabel("Age")
plt.show()
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Age of each person")
plt.show()
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.6) #alpha for trasnparecy, if there was more data and overlap
#s=df["Years_of_practice"]*60 means if a studen haa more years of practice their bubble is bigger
plt.title("Age vs years of practice")
plt.xlabel("Age")
plt.ylabel("Years of practice")
plt.show()

```

```python
import seaborn as sns #use this to create more clean and formated plots
sns.boxplot(data=df, y="Age")
plt.title("Distribution of athlete age")
plt.show
```

```python
plt.Figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.title("Box plot for age and years of practice")
plt.show
```

```python
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.show
```

```python
avg_age= df.groupby("Sport")["Years_of_practice"].mean()
avg_age.plot(kind="bar")
plt.show
```
