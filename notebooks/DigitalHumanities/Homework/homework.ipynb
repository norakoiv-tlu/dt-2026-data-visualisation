importing the dataset into this folder

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("manga.csv")

# Display the first few rows
df.head()
```

Number of volumes for each manga

```python
# Sort the data by number of volumes
volumes = df.sort_values("Volumes", ascending=True)
plt.figure
plt.barh(volumes["Manga Series"], volumes["Volumes"])
# Add labels to the axes
plt.xlabel("Number of Volumes")
plt.ylabel("Manga Series")
# Add a title
plt.title("Number of Volumes by Manga Series")
# Display the chart
plt.show()
```

The chart shows that the number of volumes varies a lot between the manga series. "JoJo's Bizarre Adventure" has the highest number of volumes in this dataset, while several series have fewer than 20 volumes. This shows that the dataset contains both very long and pretty short manga series.

Number of manga series by genre

```python
# Count how many manga series belong to each genre
genre_counts = df["Genre"].value_counts()

# Sort the values so the chart is easier to read
genre_counts = genre_counts.sort_values(ascending=True)
plt.figure
# Create a horizontal bar chart
plt.barh(genre_counts.index, genre_counts.values)
# Add axis labels
plt.xlabel("Number of Manga Series")
plt.ylabel("Genre")

# Add a title
plt.title("Number of Manga Series by Genre")
plt.show()

```

Dark Fantasy is the most common genre in the dataset. Meanwhile, superhoero and comedy series are least represented. This means that the genres are not evenly distributed, with some genres being much more represented than others.

Completion status

```python
# Count the number of manga series for each status
status_counts = df["Status"].value_counts()

# Create the figure
plt.figure

# Create a bar chart
plt.bar(status_counts.index, status_counts.values)

# Add labels to the axes
plt.xlabel("Status")
plt.ylabel("Number of Manga Series")

# Add a title
plt.title("Manga Series by Publication Status")

# Display the chart
plt.show()
```

Most of the manga series in the dataset are completed. There are fewer ongoing series, and only a small number are listed as being on hiatus. In conclusion, the dataset contains mostly completed manga.

Start year vs. number of volumes

```python
# Create a scatter plot using: start Year on the x-axis and number of Volumes on the y-axis
plt.figure

plt.scatter(df["Start Year"], df["Volumes"])

# Add labels to the axes
plt.xlabel("Start Year")
plt.ylabel("Number of Volumes")

# Add a title
plt.title("Manga Start Year vs. Number of Volumes")

# Display the chart
plt.show()
```

The scatter plot does not show a clear relationship between the start year and number of volumes. Some older manga have many volumes, but other older manga have relatively few. This suggests that the number of volumes is not only influenced by when the manga started.

Years manga was published across vs amount of volumes

```python
# Calculate how many years each manga ran
df["Years Running"] = df["End Year"] - df["Start Year"]

# Display the manga, running time, and number of volumes
df[["Manga Series", "Years Running", "Volumes"]]

# Create the scatter plot
plt.figure(figsize=(8, 6))

plt.scatter(df["Years Running"], df["Volumes"])

# Add a trend line
m, b = __import__("numpy").polyfit(df["Years Running"], df["Volumes"], 1)
plt.plot(df["Years Running"], m * df["Years Running"] + b)

# Add labels and title
plt.xlabel("Years Running")
plt.ylabel("Number of Volumes")
plt.title("Manga Running Time vs. Number of Volumes")

plt.show()
```

The scatter plot shows the relationship between how long a manga ran and how many volumes it has. There appears to be a positive relationship, meaning that manga series that ran for more years generally tend to have more volumes. However, the relationship is not perfect because some manga produced more or fewer volumes than would be expected considering their runtime.
